```python
files = [
    "../data/amazon_reviews_us_Video_Games_v1_00.tsv",
    "../data/amazon_reviews_us_Electronics_v1_00.tsv",
    "../data/amazon_reviews_us_Mobile_Electronics_v1_00.tsv"
]
```

```python
import pandas as pd

df_list = []

for file in files:
    df = pd.read_csv(
        file,
        sep="\t",
        usecols=use_cols,
        encoding="utf-8",
        on_bad_lines="skip"
    )
    
    df_list.append(df)

final_df = pd.concat(df_list, ignore_index=True)
```

```python
import re

def clean_text(text):
    text = text.lower()
    text = re.sub(r"<br\s*/?>", " ", text)
    text = re.sub(r"[^a-zA-Z ]", "", text)
    return text
```

```python
from sklearn.feature_extraction.text import ENGLISH_STOP_WORDS
```

```python
custom_stopwords = set(ENGLISH_STOP_WORDS).union({"br", "dont", "did", "does", "im", "just", "didnt", "really", "got", "doesnt"})
```

```python
def remove_stopwords(text):
    return " ".join([
        w for w in text.split()
        if w not in custom_stopwords
    ])

df_sample["text"] = df_sample["text"].apply(remove_stopwords)
```

```python
from collections import Counter

text = " ".join(
    df_sample[df_sample["is_problem"] == 1]["text"]
)

Counter(text.lower().split()).most_common(20)
```

```python
bad_words = Counter(" ".join(
    df_sample[df_sample["is_problem"] == 1]["text"]
).split())

good_words = Counter(" ".join(
    df_sample[df_sample["is_problem"] == 0]["text"]
).split())
```

```python
bad_words - good_words
```

Negatif yorumlarda sık geçen kelimeler, ürün kusurları, iade/geri ödeme şikayetleri ve teknik problemler gibi net tematik kümeler ortaya koymaktadır. Bu durum, otomatik problem yönlendirme (issue routing) sistemlerinin uygulanabilir olduğunu göstermektedir
